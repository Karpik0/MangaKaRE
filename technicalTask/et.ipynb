1. Общее описание проекта

«ЛитСфера» — это социальная платформа для чтения, совмещающая в себе функционал библиотеки, социальной сети, игровой системы (геймификация) и маркетплейса. Пользователи могут читать различные форматы контента (книги, манга, комиксы), общаться, выполнять задания и кастомизировать свой профиль.
